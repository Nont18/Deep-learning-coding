```python
import numpy as np
import random
import math
```

```python
def augment_keypoints(keypoints):
    noise = np.random.normal(0, 0.005, keypoints.shape)  # Adjust noise level
    augmented_keypoints = keypoints + noise
    # scale_factor = np.random.uniform(0.5, 1.2)
    # augmented_keypoints = keypoints * scale_factor

    theta = random.uniform(-20,20)
    theta = (math.pi * theta) / 180
    rotation_matrix = np.array([[np.cos(theta), -np.sin(theta)],
                                [np.sin(theta), np.cos(theta)]])

    all_video_frame = []
    for i in range(len(keypoints)):

        x = keypoints[i].reshape(-1,2)

        augmented_keypoints = np.array([])
        for coordinate in x:
        # print(coordinate)
            new_coordinate = np.dot(coordinate, rotation_matrix)
            # print(new_coordinate)
            # print("-----------------------------------------------------------------")
            augmented_keypoints = np.concatenate((augmented_keypoints, new_coordinate))
        
        all_video_frame.append(augmented_keypoints)

    return np.array(all_video_frame)
    # return augmented_keypoints
```
