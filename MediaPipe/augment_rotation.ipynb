### Rotating Augmentation

```python
import numpy as np
import random
import math
```

```python
x = np.random.rand(1086)
print(x)
print(x.shape)
```

```python
x = x.reshape(-1,2)
print(x)
```

```python
print(x.shape)
print(x[0])
```

```python
import random

theta = random.uniform(-20,20)
theta = (math.pi * theta) / 180
print(theta)
```

```python
rotation_matrix = [[np.cos(theta), -np.sin(theta)],
                    [np.sin(theta), np.cos(theta)]]

print(rotation_matrix)
```

```python
shift_coordinate = np.array([])
for coordinate in x:
    # print(coordinate)
    new_coordinate = np.dot(coordinate, rotation_matrix)
    print(new_coordinate)
    # print("-----------------------------------------------------------------")
    shift_coordinate = np.concatenate((shift_coordinate, new_coordinate))
print(shift_coordinate)
```

```python
print(shift_coordinate.shape)
```
